# Handwritten Digit Feature Extraction — Interactive Notebook

This notebook demonstrates the computer vision and mathematical feature extraction techniques applied to handwritten digit images.

### Pipeline Overview:
1. **Preprocessing**: Grayscale conversion, Otsu thresholding, Gaussian blur, 28x28 resizing.
2. **Edge Extraction**: Canny edge detection & Sobel gradient magnitude maps.
3. **Corner Extraction**: Harris corner response & Shi-Tomasi keypoint tracking.
4. **Intensity Statistics**: Pixel histograms, mean/std intensity, centroid location, and projection profiles.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Load sample digit image
img_path = '../dataset/mnist_samples/digit_7.png'
img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(4, 4))
plt.imshow(img, cmap='gray')
plt.title('Original Handwritten Digit (Sample 7)')
plt.axis('off')
plt.show()
```

## 1. Image Preprocessing & Otsu Binarization

```python
# Gaussian Blur & Otsu Binarization
blurred = cv2.GaussianBlur(img, (3, 3), 0)
_, binary = cv2.threshold(blurred, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
binary_28 = cv2.resize(binary, (28, 28), interpolation=cv2.INTER_AREA)

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(binary, cmap='gray')
axes[0].set_title('Full Res Binarized')
axes[1].imshow(binary_28, cmap='gray')
axes[1].set_title('Resized to 28x28 (MNIST standard)')
plt.show()
```

## 2. Edge Detection (Canny vs Sobel Operator)

```python
# Canny Edges
canny_edges = cv2.Canny(binary_28, 50, 150)

# Sobel Gradient Magnitude
sobel_x = cv2.Sobel(binary_28, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(binary_28, cv2.CV_64F, 0, 1, ksize=3)
sobel_mag = np.sqrt(sobel_x**2 + sobel_y**2)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(canny_edges, cmap='magma')
axes[0].set_title(f'Canny Edge Map (Pixels: {np.count_nonzero(canny_edges)})')
axes[1].imshow(sobel_mag, cmap='viridis')
axes[1].set_title('Sobel Gradient Magnitude Map')
plt.show()
```

## 3. Corner Detection (Harris & Shi-Tomasi)

```python
# Harris Corners
img_float = np.float32(binary_28)
harris_resp = cv2.cornerHarris(img_float, 2, 3, 0.04)
harris_corners = harris_resp > (0.01 * harris_resp.max())

# Shi-Tomasi Corners
corners = cv2.goodFeaturesToTrack(binary_28, maxCorners=20, qualityLevel=0.1, minDistance=3)

# Visualization
color_img = cv2.cvtColor(cv2.resize(binary_28, (280, 280), interpolation=cv2.INTER_NEAREST), cv2.COLOR_GRAY2BGR)
if corners is not None:
    for pt in corners:
        x, y = pt[0]
        cx, cy = int((x + 0.5) * 10), int((y + 0.5) * 10)
        cv2.circle(color_img, (cx, cy), 5, (0, 0, 255), -1)

plt.figure(figsize=(5, 5))
plt.imshow(cv2.cvtColor(color_img, cv2.COLOR_BGR2RGB))
plt.title(f'Shi-Tomasi Corners Overlaid ({len(corners) if corners is not None else 0} keypoints)')
plt.axis('off')
plt.show()
```

## 4. Intensity Histogram & Projection Profiles

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# 1. Pixel Intensity Histogram
axes[0].hist(binary_28.flatten(), bins=16, color='purple', edgecolor='white')
axes[0].set_title('Pixel Intensity Histogram')

# 2. Horizontal Projection
horiz_proj = np.sum(binary_28, axis=1)
axes[1].plot(horiz_proj, range(28), color='teal', linewidth=2)
axes[1].invert_yaxis()
axes[1].set_title('Horizontal Projection (Row Sums)')

# 3. Vertical Projection
vert_proj = np.sum(binary_28, axis=0)
axes[2].plot(range(28), vert_proj, color='orange', linewidth=2)
axes[2].set_title('Vertical Projection (Column Sums)')

plt.tight_layout()
plt.show()
```
